```python
import cv2
import matplotlib.pyplot as plt
import os
```

here we will be working on retina data from the dataset - https://www.kaggle.com/datasets/andrewmvd/drive-digital-retinal-images-for-vessel-extraction

```python
# Path to dataset
img_path = "data/training/images/21_training.tif"
gt_path = "data/training/1st_manual/21_manual1.gif"

# Load images
img = cv2.imread(img_path)
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

gt = cv2.imread(gt_path, 0)
```

```python
# Filtering
blur = cv2.GaussianBlur(img_gray, (5,5), 0)

# Segmentation (Adaptive Threshold)
thresh = cv2.adaptiveThreshold(
    blur, 255,
    cv2.ADAPTIVE_THRESH_MEAN_C,
    cv2.THRESH_BINARY,
    11, 2
)

# Edge detection (optional)
edges = cv2.Canny(blur, 50, 150)
```

```python
# Plot results
plt.figure(figsize=(12,6))

plt.subplot(2,3,1)
plt.title("Original")
plt.imshow(img)
plt.axis('off')

plt.subplot(2,3,2)
plt.title("Grayscale")
plt.imshow(img_gray, cmap='gray')
plt.axis('off')

plt.subplot(2,3,3)
plt.title("Filtered")
plt.imshow(blur, cmap='gray')
plt.axis('off')

plt.subplot(2,3,4)
plt.title("Segmented")
plt.imshow(thresh, cmap='gray')
plt.axis('off')

plt.subplot(2,3,5)
plt.title("Edges")
plt.imshow(edges, cmap='gray')
plt.axis('off')

plt.subplot(2,3,6)
plt.title("Ground Truth")
plt.imshow(gt, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```
